```python
# ==============================================================================
# IMPLEMENTASI SEMINAR 5: HUMAN-COMPUTER INTERACTION & DATA VISUALIZATION
# Topik: Interactive Data Dashboard dengan Prinsip HCI & Shneiderman's Mantra
# ==============================================================================

import pandas as pd
import numpy as np
import plotly.express as px

print("Memuat Data untuk Dashboard Interaktif...\n")

# 1. GENERASI DATASET OPERASIONAL (Analisis Kinerja Jaringan Wi-Fi Bulanan)
np.random.seed(42)
n_data = 300

df_dashboard = pd.DataFrame({
    'Bulan': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun'], n_data),
    'Wilayah': np.random.choice(['Bengkulu Utara', 'Bengkulu Selatan', 'Kota Bengkulu', 'Rejang Lebong'], n_data),
    'Paket_Layanan': np.random.choice(['Basic 10Mbps', 'Family 30Mbps', 'Pro 100Mbps'], n_data),
    'Penggunaan_GB': np.random.normal(150, 40, n_data),
    'Kepuasan_Pelanggan': np.random.randint(1, 6, n_data) # Skala 1 - 5
})

# 2. IMPLEMENTASI "OVERVIEW FIRST" (Ringkasan Metrik Utama / KPI Cards)
total_pengguna = len(df_dashboard)
rata_penggunaan = df_dashboard['Penggunaan_GB'].mean()
rata_kepuasan = df_dashboard['Kepuasan_Pelanggan'].mean()

print("=== SHNEIDERMAN'S MANTRA: OVERVIEW (KPI METRICS) ===")
print(f"Total Sampel Pengguna : {total_pengguna}")
print(f"Rata-rata Penggunaan  : {rata_penggunaan:.2f} GB")
print(f"Skor Kepuasan Rata2   : {rata_kepuasan:.2f} / 5.0\n")

# 3. IMPLEMENTASI "ZOOM & FILTER" SERTA VISUALISASI BERSIH (MINIMAL CHARTJUNK)
# Menggunakan Plotly untuk grafik interaktif yang mendukung hover, zoom, dan filter visual
fig = px.box(
    df_dashboard,
    x='Wilayah',
    y='Penggunaan_GB',
    color='Paket_Layanan',
    title="Distribusi Penggunaan Data per Wilayah & Paket (Interactive Boxplot)",
    labels={'Penggunaan_GB': 'Penggunaan Data (GB)', 'Wilayah': 'Wilayah Operasional'}
)

# Menerapkan tata letak bersih sesuai prinsip HCI (mengurangi beban kognitif pengguna)
fig.update_layout(
    plot_bgcolor='white',
    paper_bgcolor='white',
    font=dict(family="Arial", size=12, color="#333333")
)
fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='#eeeeee')
fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='#eeeeee')

# Tampilkan grafik interaktif
fig.show()
```
